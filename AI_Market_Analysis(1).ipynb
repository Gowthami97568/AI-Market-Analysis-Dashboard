```python
import pandas as pd

# get file name automatically
file_name = list(uploaded.keys())[0]

# load dataset
df = pd.read_csv(file_name)

# show data
df.head()
```

```python
# Check missing values
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

# Remove duplicates
df = df.drop_duplicates()

# Reset index
df = df.reset_index(drop=True)

print("Data cleaned ✅")
```

```python
top_countries = df.groupby('Country')['AI_Investment_BillionUSD'].sum().sort_values(ascending=False)

top_countries.head(10)
```

```python
yearly_trend = df.groupby('Year')['AI_Investment_BillionUSD'].sum()

yearly_trend
```

```python
# Yearly growth rate
yearly_growth = yearly_trend.pct_change() * 100

yearly_growth
```

```python
# Correlation between variables
correlation = df[['AI_Investment_BillionUSD',
                  'Productivity_Index',
                  'Employment_Rate_Percent']].corr()

correlation
```

```python
import matplotlib.pyplot as plt

# Plot AI investment trend
yearly_trend.plot(kind='line', title='AI Investment Growth')

plt.show()
```

```python
df.to_csv("cleaned_ai_data.csv", index=False)
```
